```python
import os
from IPython.display import Image, display
import pandas as pd
import aoristic
```

### 1. Use function with EU formatted dates (yyyy-mm-dd)

Edited dataset for Washington DC burglaries, with EU formatted date time columns added:

 * `comm_date_from`
 * `comm_date_to`
 * `comm_time_fr`
 * `comm_time_to`

```python
# Test EU date time formatted data
url = 'https://github.com/routineactivity/adhoc_notebooks/blob/main/aoristic/data/dc_burglary_edited.csv?raw=true'
eu_dttm = pd.read_csv(url)
eu_dttm.columns
```

Check `aoristic.aoristic` to view the inputs

```python
aoristic.aoristic
```

```python
# Generate aoristic plots for the dataset
aoristic.aoristic(
    input_df=eu_dttm,
    comm_date_fr='comm_date_from',
    comm_date_to='comm_date_to',
    comm_time_fr='comm_time_fr',
    comm_time_to='comm_time_to',
    output_directory='outputs/eu_format_results'
)
```

Results can be retrieved to display in notebook, or used elsewhere.

```python
# View results
display(Image(filename='outputs/eu_format_results/aoristic_heatmap.png'))
display(Image(filename='outputs/eu_format_results/aoristic_totals_by_day.png'))
display(Image(filename='outputs/eu_format_results/aoristic_totals_by_hour.png'))
```

### 2. Use function with US formatted dates (yyyy-dd-mm / mm-dd-yyyy)

Dataset for NYPD shootings from NYC open data, with US formatted date time. In this dataset there is only a single date and time of occurrence, hence is duplicated when using the function.

 * `OCCUR_DATE`
 * `OCCUR_TIME`

```python
# Test US date time formatted data
url = 'https://github.com/routineactivity/adhoc_notebooks/blob/main/aoristic/data/NYPD_Shooting_Incident_Data__Year_To_Date__20250406.csv?raw=true'
us_dttm = pd.read_csv(url)
us_dttm.columns
```

```python
aoristic.aoristic_us
```

```python
aoristic.aoristic_us(
    input_df=us_dttm,
    comm_date_fr='OCCUR_DATE',
    comm_date_to='OCCUR_DATE',
    comm_time_fr='OCCUR_TIME',
    comm_time_to='OCCUR_TIME',
    output_directory='outputs/us_format_results'
)
```

```python
display(Image(filename='outputs/us_format_results/aoristic_heatmap.png'))
display(Image(filename='outputs/us_format_results/aoristic_totals_by_day.png'))
display(Image(filename='outputs/us_format_results/aoristic_totals_by_hour.png'))
```
